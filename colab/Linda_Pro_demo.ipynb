```
from linda_pro import LindaPro
text = """Paste a text of at least a few hundred words here."""
for mode in ('sensitive', 'precise'):
    r = LindaPro(mode=mode).detect([text])[0]
    print(mode, r['verdict'], 'ai_share=%.0f%%' % (100 * r['ai_share']), 'windows', r['n_windows'])
```
